```python
# Se seleccionan del dataset de dinagua el caudal y nivel de agua de la estación Florida (Puente Ruta 5)

idx = pd.IndexSlice
est = "Florida (Puente Ruta 5)"

data_dinagua = data_dinagua.loc[:, idx[["Caudal (Q) [m3/s]", "Nivel de agua (h) [m]"], [est]]]
data_dinagua.head()
```

```python
cols = ["Florida","Cerro Colorado","La Cruz","Mendoza","Reboledo","San Gabriel",
        "Sarandi Grande","Villa 25 de Mayo","Villa Cardal"]

data_inumet = data_inumet.loc[:, data_inumet.columns.get_level_values(1).isin(cols)].copy()

data_inumet.head()
```

```python
# DATASET DINAGUA: se realiza el pasaje a valores promedios diarios mensuales (un valor por mes)

dinagua_mensual = data_dinagua.resample('ME').mean()

# Se ajusta el índice para asignar la fecha al último día de cada mes

dinagua_mensual.index = dinagua_mensual.index.to_period('M').to_timestamp('M')

dinagua_mensual.head()
```

```python
# DATASET INUMET: se realiza el pasaje a valores promedios diarios mensuales (mes a fin de mes)

inumet_mensual = data_inumet.resample('ME').mean()

inumet_mensual.index = inumet_mensual.index.to_period('M').to_timestamp('M')

inumet_mensual.head()
```

```python
# DATASET INIA: se realiza el pasaje a valores promedios diarios mensuales (un valor por mes)

inia_mensual = data_inia.resample('ME').mean()

# Se ajusta el índice para asignar la fecha al último día de cada mes

inia_mensual.index = inia_mensual.index.to_period('M').to_timestamp('M')

inia_mensual.head()
```

```python
# Se asegura que todos los sets tengan el mismo índice de fecha para la posterior unión entre ellos

data_calidad.index = pd.to_datetime(data_calidad.index)
dinagua_mensual.index = pd.to_datetime(dinagua_mensual.index)
inumet_mensual.index = pd.to_datetime(inumet_mensual.index)
inia_mensual.index = pd.to_datetime(inia_mensual.index)

# Se combinan los datasets

dataset_completo = pd.concat([data_calidad, dinagua_mensual, inumet_mensual, inia_mensual], axis=1)

dataset_completo.head()
```

```python
dataset_completo = dataset_completo.loc["2014-01-01":"2022-12-31"].copy()

dataset_completo.tail(5)
```

```python
# Se cambia la estructura del dataset

dataset_completo.columns = [' '.join(col).strip() for col in dataset_completo.columns.values]
dataset_completo.head()
```

```python
# Se obtiene un dataset mensual de extremos de caudal

dinagua_mensual_extremos = (
    data_dinagua[[('Caudal (Q) [m3/s]', 'Florida (Puente Ruta 5)')]]
    .resample('ME')
    .agg(['max', 'min'])
)

# Se ajusta el índice al último día de cada mes
dinagua_mensual_extremos.index = dinagua_mensual_extremos.index.to_period('M').to_timestamp('M')

dinagua_mensual_extremos.columns = [
    'Caudal Maximo (Qmax) [m3/s] Florida (Puente Ruta 5)',
    'Caudal Minimo (Qmin) [m3/s] Florida (Puente Ruta 5)'
]

# Se selecciona el intervalo de trabajo
dinagua_mensual_extremos = dinagua_mensual_extremos.loc["2014-01-01":"2022-12-31"]

dinagua_mensual_extremos.head()
```

```python
# Se obtiene un dataset mensual de precipitaciones máximas

inumet_mensual_extremos = (
    data_inumet
    .resample('ME')
    .max()
)

# Se ajusta el índice al último día de cada mes
inumet_mensual_extremos.index = inumet_mensual_extremos.index.to_period('M').to_timestamp('M')

inumet_mensual_extremos.columns = [
    f'Precipitacion maxima (Pmax) [mm] {station}'
    for variable, station in inumet_mensual_extremos.columns
]

# Se selecciona el intervalo de trabajo
inumet_mensual_extremos = inumet_mensual_extremos.loc["2014-01-01":"2022-12-31"]

inumet_mensual_extremos.head()
```

```python
# Se unen a dataset_completo los datasets de valores máximos de caudal y precipitación

dataset_completo.index = pd.to_datetime(dataset_completo.index)
dinagua_mensual_extremos.index = pd.to_datetime(dinagua_mensual_extremos.index)
inumet_mensual_extremos.index = pd.to_datetime(inumet_mensual_extremos.index)

dataset_completo = pd.concat(
    [
        dataset_completo,
        dinagua_mensual_extremos,
        inumet_mensual_extremos
    ],
    axis=1
)

dataset_completo.head()
```

# INTERESA CALCULAR EL ÍNDICE DE ESTADO TRÓFICO EN LAS ESTACIONES:

## EPSE040 (PUNTO DE CÁLCULO DEL ÍNDICE DE ESTADO TRÓFICO PARA SISTEMA LÉNTICO)
## XSLH030 (PUNTO DE CÁLCULO DEL ÍNDICE DE ESTADO TRÓFICO PARA SISTEMA LÓTICO)

# Algunas notas sobre el IET

### DINACEA calcula el IET de Lamparelli para ríos, y para lagos el IET de Cunha, aunque en otros documentos (por ejemplo, el Informe del estado del ambiente 2022-2024) se hace referencia a los niveles guía  definidos en "ESTABLECIMIENTO DE NIVELES GUÍA DE INDICADORES DE ESTADO TRÓFICO EN CUERPOS DE AGUA SUPERFICIALES" 

### El IET de Lamparelli en realidad se calcula anualmente: primero se calcula la media geométrica de las concentraciones mensuales de PT, y con ese valor se calcula IET(PT) para ese año. En forma análoga, se calcula la media geométrica de las concentraciones mensuales de Cl, y con ese valor se calcula IET(Cl) para ese año. Finalmente, se calcula el promedio aritmético de IET(PT) y IET(Cl) para obtener el IET anual

### Si sólo se tiene uno de los dos parámetros, se procede a utilizar el IET correspondiente a ese parámetro

### Si bien, como se dijo, el IET se calcula año a año, es válido obtener los valores mensuales para conocer la variación estacional del IET
### Referencia de lo anterior: Apêndice D - Índices de Qualidade das Águas, Critérios de Avaliação da Qualidade dos Sedimentos e Indicador de Controle de Fontes 2022
### En el cálculo mensual, en condiciones ambientales sin restricciones, el IET(cl) y el IET(P) deberían coincidir. En ausencia de luz o temperatura, por ejemplo, el fitoplancton no se desarrollará plenamente para la cantidad de nutrientes existentes

### En el IET de Lamparelli para ríos, se utiliza sólo el fósforo, y se aplica la media geométrica de las concentraciones, como ya se dijo
### En el IET de Cunha para embalses, se utiliza el promedio aritmético de los IET


#### A continuación se presentan las fórmulas utilizadas para el cálculo del Índice de Estado Trófico, a partir del fósforo y de la clorofila-a, tanto para cuerpos lóticos como lénticos<br>

<div align="center">################################################################################</div>

<div align="center">

# Lamparelli (2004)

</div>

<div align="center">
    
### Índice de Estado Trófico (IET) para cuerpos lóticos
    
</div>

$$
IET(PT) = 10 \left[6 - \frac{0.42 - 0.36 \cdot \ln(PT)}{\ln(2)} \right] - 20
$$

$$
IET(Cl) = 10 \left[6 - \frac{-0.7 - 0.6 \cdot \ln(Cl)}{\ln(2)} \right] - 20
$$

$$
IET = \frac{IET(PT) + IET(Cl)}{2}
$$

<div align="center">
    
### Índice de Estado Trófico (IET) para cuerpos lénticos
    
</div>

$$
IET(PT) = 10 \left[6 - \frac{1.77 - 0.42 \cdot \ln(PT)}{\ln(2)} \right] 
$$

$$
IET(Cl) = 10 \left[6 - \frac{0.92 - 0.34 \cdot \ln(Cl)}{\ln(2)} \right] 
$$

$$
IET = \frac{IET(PT) + IET(Cl)}{2}
$$

<div align="center">
IET(PT) es el índice de estado trófico determinado a partir del fósforo total;<br>
IET(Cl) es el índice de estado trófico determinado a partir de la clorofila;<br>
PT es la concentración de fósforo total medida en [µg/L];<br> 
Cl es la concentración de clorofila medida en [µg/L];<br>
</div><br>    
 
<div align="center">################################################################################</div>

<div align="center">

# Cunha (2013)

</div>

<div align="center">
    
### Índice de Estado Trófico (IET) para cuerpos lénticos
    
</div>

$$
IET(PT) = 10\left[\,6 - \frac{-0{,}27637\,\ln(PT) + 1{,}329766}{\ln(2)}\right]
$$

$$
IET(Cl) = 10\left[\,6 - \frac{-0{,}2512\,\ln(Cl) + 0{,}842257}{\ln(2)}\right]
$$

$$
IET = \frac{IET(PT) + IET(Cl)}{2}
$$

```python
import numpy as np
import pandas as pd

PT_lotico = "Fósforo Total (PT) [μg/L] XSLH030"
PT_lentico = "Fósforo Total (PT) [μg/L] EPSE040"

estado_trofico = pd.DataFrame(index=dataset_completo.index)

estado_trofico["IET(PT) Lótico [XSLH030]"] = 10 * (6 - ((0.42 - 0.36 * np.log(dataset_completo[PT_lotico])) / np.log(2))) - 20

estado_trofico["IET(PT) Léntico [EPSE040]"] = 10 * (6 - ((-0.27637 * np.log(dataset_completo[PT_lentico]) + 1.329766) / np.log(2)))

estado_trofico.head(5)
```

```python
tabla_estadisticos_iet = pd.DataFrame({
    "Min": estado_trofico.min(),
    "P25": estado_trofico.quantile(0.25),
    "Mediana": estado_trofico.median(),
    "Media": estado_trofico.mean(),
    "P75": estado_trofico.quantile(0.75),
    "Max": estado_trofico.max(),
    "Desv_std": estado_trofico.std()
})

tabla_estadisticos_iet = tabla_estadisticos_iet.round(2)

tabla_estadisticos_iet.to_csv(
    "estadisticos_IET_respuesta_previo.csv",
    encoding="utf-8-sig"
)

tabla_estadisticos_iet
```
