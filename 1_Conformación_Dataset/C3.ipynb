```python
# lectura del documento dataset_completo.parquet creado en el cuaderno C1

path1 = './0-Datasets/dataset_completo.parquet'
dataset_completo = pd.read_parquet(path1)

dataset_completo.head()
```

```python
# lectura del documento dataset_completo_crudo.parquet creado en el cuaderno C2

path2 = './0-Datasets/dataset_completo_crudo.parquet'
dataset_completo_crudo = pd.read_parquet(path2)

dataset_completo_crudo.head()
```

```python
# Se crea un dataset con valores de fósforo, para datasets imputado y crudo

cols = [
    "Fósforo Total (PT) [μg/L] EPSE040",
    "Fósforo Total (PT) [μg/L] XSLH030",
]

dataset_fosforo = pd.DataFrame(index=dataset_completo.index)

ordered_cols = []
for c in cols:
    dataset_fosforo[c] = dataset_completo[c]
    dataset_fosforo[f"{c} [crudo]"] = dataset_completo_crudo[c]
    ordered_cols.extend([c, f"{c} [crudo]"])

dataset_fosforo = dataset_fosforo.loc[:, ordered_cols]

dataset_fosforo.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

sns.set_theme(style="whitegrid", context="talk")

# Ajustes 
alpha_crudo = 0.6   # defino la transparencia de puntos (0 a 1)
size_crudo  = 140    # defino el tamaño de puntos

def plot_pt_estacion(dataset_fosforo, estacion):
    col_full = f"Fósforo Total (PT) [μg/L] {estacion}"
    col_crudo = f"{col_full} [crudo]"

    df_plot = dataset_fosforo[[col_full, col_crudo]].reset_index()

    plt.figure(figsize=(14, 6))

    sns.lineplot(
        data=df_plot, x="date", y=col_full,
        linewidth=2.5, label=f"{estacion} (imputado)"
    )

    sns.scatterplot(
        data=df_plot, x="date", y=col_crudo,
        s=size_crudo, alpha=alpha_crudo, color="red",
        label=f"{estacion} (crudo)"
    )

    plt.title(f"Fósforo Total (PT) — {estacion} (imputado vs crudo)")
    plt.xlabel("Fecha (fin de mes)")
    plt.ylabel("PT [μg/L]")
    plt.legend(title="")
    plt.tight_layout()
    plt.show()

# EPSE020
plot_pt_estacion(dataset_fosforo, "EPSE040")

# XSLH030
plot_pt_estacion(dataset_fosforo, "XSLH030")
```

```python
import numpy as np
import pandas as pd

# --- columnas en dataset_fosforo ---
PT_lotico_imp   = "Fósforo Total (PT) [μg/L] XSLH030"
PT_lotico_crudo = "Fósforo Total (PT) [μg/L] XSLH030 [crudo]"

PT_lentico_imp   = "Fósforo Total (PT) [μg/L] EPSE040"
PT_lentico_crudo = "Fósforo Total (PT) [μg/L] EPSE040 [crudo]"

# --- helper para log seguro ---
def safe_ln(s):
    s = pd.to_numeric(s, errors="coerce")
    s = s.where(s > 0)            # PT <= 0 -> NaN
    return np.log(s)

ln2 = np.log(2)

trofico_imputadoycrudo = pd.DataFrame(index=dataset_fosforo.index)

# =========================
# LÓTICO (Lamparelli) - solo IET(PT)
# IET(PT) = 10 * (6 - ((0.42 - 0.36 ln(PT))/ln2)) - 20
# =========================
trofico_imputadoycrudo["IET(PT) Lótico_Lamparelli [XSLH030]"] = (
    10 * (6 - ((0.42 - 0.36 * safe_ln(dataset_fosforo[PT_lotico_imp])) / ln2)) - 20
)

trofico_imputadoycrudo["IET(PT) Lótico_Lamparelli [XSLH030] [crudo]"] = (
    10 * (6 - ((0.42 - 0.36 * safe_ln(dataset_fosforo[PT_lotico_crudo])) / ln2)) - 20
)

# =========================
# LÉNTICO (Cunha) - solo IET(PT)
# IET(PT) = 10 * [ 6 - ( (-0.27637 ln(PT) + 1.329766) / ln2 ) ]
# =========================
trofico_imputadoycrudo["IET(PT) Léntico_Cunha [EPSE040]"] = (
    10 * (6 - ((-0.27637 * safe_ln(dataset_fosforo[PT_lentico_imp]) + 1.329766) / ln2))
)

trofico_imputadoycrudo["IET(PT) Léntico_Cunha [EPSE040] [crudo]"] = (
    10 * (6 - ((-0.27637 * safe_ln(dataset_fosforo[PT_lentico_crudo]) + 1.329766) / ln2))
)

trofico_imputadoycrudo.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

sns.set_theme(style="whitegrid", context="talk")

# Ajustes
alpha_crudo = 0.6   # transparencia puntos (0 a 1)
size_crudo  = 140   # tamaño puntos

def plot_iet(dataset_trofico, col_full, col_crudo, titulo, ylabel="IET"):
    df_plot = dataset_trofico[[col_full, col_crudo]].reset_index()

    plt.figure(figsize=(14, 6))

    # línea continua (imputado/procesado)
    sns.lineplot(
        data=df_plot, x="date", y=col_full,
        linewidth=2.5, label="Imputado"
    )

    # puntos (crudo)
    sns.scatterplot(
        data=df_plot, x="date", y=col_crudo,
        s=size_crudo, alpha=alpha_crudo, color="red",
        label="Crudo"
    )

    plt.title(titulo)
    plt.xlabel("Fecha (fin de mes)")
    plt.ylabel(ylabel)
    plt.legend(title="")
    plt.tight_layout()
    plt.show()


# 1) IET(PT) Léntico Cunha [EPSE040]
plot_iet(
    trofico_imputadoycrudo,
    col_full="IET(PT) Léntico_Cunha [EPSE040]",
    col_crudo="IET(PT) Léntico_Cunha [EPSE040] [crudo]",
    titulo="IET(PT) Léntico (Cunha) — EPSE040 (Imputado vs Crudo)"
)

# 2) IET(PT) Lótico Lamparelli [XSLH030]
plot_iet(
    trofico_imputadoycrudo,
    col_full="IET(PT) Lótico_Lamparelli [XSLH030]",
    col_crudo="IET(PT) Lótico_Lamparelli [XSLH030] [crudo]",
    titulo="IET(PT) Lótico (Lamparelli) — XSLH030 (Imputado vs Crudo)"
)

```

```python
import numpy as np
import pandas as pd

PT_lotico  = "Fósforo Total (PT) [μg/L] XSLH030"
PT_lentico = "Fósforo Total (PT) [μg/L] EPSE040"

ln2 = np.log(2)

def safe_ln(s):
    s = pd.to_numeric(s, errors="coerce")
    s = s.where(s > 0)
    return np.log(s)

def add_iet_pt_lotico_lamparelli(df):
    df["IET(PT) Lótico_Lamparelli [XSLH030]"] = (
        10 * (6 - ((0.42 - 0.36 * safe_ln(df[PT_lotico])) / ln2)) - 20
    )
    return df

def add_iet_pt_lentico_cunha(df):
    df["IET(PT) Léntico_Cunha [EPSE040]"] = (
        10 * (6 - ((-0.27637 * safe_ln(df[PT_lentico]) + 1.329766) / ln2))
    )
    return df

########################## Se crean nuevos datasets con el agregado de los IET ########################################
dataset_completo_ok_iet = add_iet_pt_lentico_cunha(
    add_iet_pt_lotico_lamparelli(dataset_completo_ok.copy())
)

dataset_completo_crudo_iet = add_iet_pt_lentico_cunha(
    add_iet_pt_lotico_lamparelli(dataset_completo_crudo.copy())
)

# chequeo rápido
dataset_completo_ok_iet[[PT_lotico, PT_lentico,
                         "IET(PT) Lótico_Lamparelli [XSLH030]",
                         "IET(PT) Léntico_Cunha [EPSE040]"]].head()
```

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Estilo
sns.set_theme(style="darkgrid", context="notebook")

def boxplot_comparacion(df_imp, df_crudo, col_imp, col_crudo, titulo, ylabel, filename):
    df_plot = pd.DataFrame({
        "Imputado": pd.to_numeric(df_imp[col_imp], errors="coerce"),
        "Crudo": pd.to_numeric(df_crudo[col_crudo], errors="coerce")
    }).melt(var_name="Tipo", value_name="Valor").dropna()

    plt.figure(figsize=(6, 4))

    ax = sns.boxplot(
        data=df_plot,
        x="Tipo",
        y="Valor",
        width=0.45,
        boxprops={"alpha": 0.9},
        whiskerprops={"linewidth": 1.5},
        medianprops={"color": "black", "linewidth": 1.8}
    )

    sns.stripplot(
        data=df_plot,
        x="Tipo",
        y="Valor",
        alpha=0.45,
        size=3,
        jitter=0.15,
        color="black"
    )

    ax.set_title(titulo, pad=10, weight="bold")
    ax.set_xlabel("")
    ax.set_ylabel(ylabel)
    ax.set_facecolor("#EAEAF2")

    plt.tight_layout()
    plt.savefig(os.path.join("figuras", filename), dpi=300, bbox_inches="tight")
    plt.show()


# =========================================
# 1) Fósforo - XSLH030
# =========================================
boxplot_comparacion(
    df_imp=dataset_completo_ok_iet,
    df_crudo=dataset_completo_crudo_iet,
    col_imp="Fósforo Total (PT) [μg/L] XSLH030",
    col_crudo="Fósforo Total (PT) [μg/L] XSLH030",
    titulo="Fósforo Total (PT) — XSLH030",
    ylabel="PT [μg/L]",
    filename="boxplot_PT_XSLH030.png"
)

# =========================================
# 2) Fósforo - EPSE040
# =========================================
boxplot_comparacion(
    df_imp=dataset_completo_ok_iet,
    df_crudo=dataset_completo_crudo_iet,
    col_imp="Fósforo Total (PT) [μg/L] EPSE040",
    col_crudo="Fósforo Total (PT) [μg/L] EPSE040",
    titulo="Fósforo Total (PT) — EPSE040",
    ylabel="PT [μg/L]",
    filename="boxplot_PT_EPSE040.png"
)

# =========================================
# 3) IET - XSLH030
# =========================================
boxplot_comparacion(
    df_imp=dataset_completo_ok_iet,
    df_crudo=dataset_completo_crudo_iet,
    col_imp="IET(PT) Lótico_Lamparelli [XSLH030]",
    col_crudo="IET(PT) Lótico_Lamparelli [XSLH030]",
    titulo="IET(PT) Lótico — XSLH030",
    ylabel="IET(PT)",
    filename="boxplot_IET_XSLH030.png"
)

# =========================================
# 4) IET - EPSE040
# =========================================
boxplot_comparacion(
    df_imp=dataset_completo_ok_iet,
    df_crudo=dataset_completo_crudo_iet,
    col_imp="IET(PT) Léntico_Cunha [EPSE040]",
    col_crudo="IET(PT) Léntico_Cunha [EPSE040]",
    titulo="IET(PT) Léntico — EPSE040",
    ylabel="IET(PT)",
    filename="boxplot_IET_EPSE040.png"
)
```

```python
import os
import math
import textwrap
import matplotlib.pyplot as plt

# =========================================================
# CONFIGURACION GENERAL
# =========================================================

# Carpeta de salida
output_dir = "Figuras"
os.makedirs(output_dir, exist_ok=True)

# Variables a excluir
exclude_patterns = [
    "Caudal Maximo (Qmax)",
    "Caudal Minimo (Qmin)",
    "Precipitacion maxima (Pmax)",
    "IET(PT) Lótico_Lamparelli [XSLH030]",
    "IET(PT) Léntico_Cunha [EPSE040]"
]

# Copia del dataset excluyendo las variables indicadas
df = dataset_completo_ok_iet.copy()

df = df[
    [
        c for c in df.columns
        if not any(pat in c for pat in exclude_patterns)
    ]
]

# Orden deseado de estaciones de calidad de agua
quality_stations = [
    "EPSE020",
    "EPSE040",
    "EPSE030",
    "XSLH030",
    "XSLH020",
    "XSLH010"
]

# Estaciones INUMET
inumet_stations = [
    "Cerro Colorado",
    "Florida",
    "La Cruz",
    "Mendoza",
    "Reboledo",
    "San Gabriel",
    "Sarandi Grande",
    "Villa 25 de Mayo",
    "Villa Cardal"
]

# Orden deseado para variables de calidad de agua
quality_var_order = [
    "Fósforo Total (PT) [μg/L]",
    "Nitrógeno Total (NT) [mg/L]",
    "Ion Nitrato (NO3-) [mg/L]",
    "Ion Nitrito (NO2-) [mg/L]",
    "Ion Amonio (NH4+) [mg/L]",
    "Ion Fosfato (PO4-) [µg/L]",
    "Sólidos totales (ST) [mg/L]",
    "Sólidos suspendidos totales (SST) [mg/L]",
    "Turbidez [NTU]",
    "Temperatura del agua (T) [ºC]",
    "Oxigeno Disuelto (OD) [mg/L]",
    "Demanda Bioquímica de Oxigeno (DBO) [mg/L]",
    "Clorofila-a (Chl-a) [μg/L]",
    "Potencial de Hidrogeno (pH) [NA]",
    "Conductividad [μS/cm]",
    "Glifosato [µg/L]"
]

# =========================================================
# FUNCIONES AUXILIARES
# =========================================================

def style_axis(ax):
    """
    Estilo tipo seaborn clásico:
    fondo gris claro, grilla blanca y sin bordes visibles.
    """
    ax.set_facecolor("#EAEAF2")

    ax.grid(
        True,
        which="major",
        axis="both",
        color="white",
        linewidth=1.0
    )

    ax.set_axisbelow(True)

    for spine in ax.spines.values():
        spine.set_visible(False)


def wrap_title(text, width=32):
    return "\n".join(textwrap.wrap(text, width=width))


def parse_quality_columns(columns, stations):
    """
    Separa el nombre de la variable de la estación.

    Ejemplo:
    'Fósforo Total (PT) [μg/L] EPSE040'

    pasa a:
    variable = 'Fósforo Total (PT) [μg/L]'
    estación = 'EPSE040'
    """
    out = {}

    for col in columns:
        for st in stations:

            suffix = f" {st}"

            if col.endswith(suffix):

                var_base = col[:-len(suffix)].strip()

                out.setdefault(var_base, {})[st] = col

                break

    return out


# Configuración de los valores atípicos
flier_style = dict(
    marker=".",
    markersize=3,
    markeredgecolor="black",
    markerfacecolor="black",
    linestyle="none"
)

# =========================================================
# 1) BOXPLOTS DE CALIDAD DE AGUA
# =========================================================

quality_cols = [
    c for c in df.columns
    if any(c.endswith(f" {st}") for st in quality_stations)
]

quality_map = parse_quality_columns(
    quality_cols,
    quality_stations
)

# Mantener solo las variables que realmente estén presentes
quality_vars_present = [
    v for v in quality_var_order
    if v in quality_map
]

n_vars = len(quality_vars_present)

ncols = 3
nrows = math.ceil(n_vars / ncols)

fig, axes = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(15, 3.8 * nrows)
)

axes = axes.flatten()

for i, var in enumerate(quality_vars_present):

    ax = axes[i]

    station_dict = quality_map[var]

    stations_present = [
        st for st in quality_stations
        if st in station_dict
    ]

    data = [
        df[station_dict[st]].dropna().values
        for st in stations_present
    ]

    ax.boxplot(
        data,
        tick_labels=stations_present,
        widths=0.6,
        patch_artist=False,
        showfliers=True,
        flierprops=flier_style
    )

    # Identificación de la variable de cada panel
    ax.set_title(
        wrap_title(var, width=28),
        fontsize=9
    )

    ax.tick_params(
        axis="x",
        labelrotation=45,
        labelsize=8
    )

    ax.tick_params(
        axis="y",
        labelsize=8
    )

    style_axis(ax)

# Apagar ejes sobrantes
for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.tight_layout()

quality_path = os.path.join(
    output_dir,
    "boxplots_calidad_agua.png"
)

plt.savefig(
    quality_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================================================
# 2) BOXPLOTS HIDROLOGICOS
# =========================================================

hydro_order = [
    "Caudal (Q) [m3/s] Florida (Puente Ruta 5)",
    "Nivel de agua (h) [m] Florida (Puente Ruta 5)"
]

hydro_cols = [
    c for c in hydro_order
    if c in df.columns
]

fig, axes = plt.subplots(
    1,
    len(hydro_cols),
    figsize=(5.5 * len(hydro_cols), 4)
)

if len(hydro_cols) == 1:
    axes = [axes]

for ax, col in zip(axes, hydro_cols):

    data = df[col].dropna().values

    ax.boxplot(
        [data],
        tick_labels=["Florida"],
        widths=0.5,
        patch_artist=False,
        showfliers=True,
        flierprops=flier_style
    )

    short_name = col.replace(
        " Florida (Puente Ruta 5)",
        ""
    )

    # Identificación de cada variable
    ax.set_title(
        wrap_title(short_name, width=28),
        fontsize=9
    )

    ax.tick_params(
        axis="x",
        labelsize=8
    )

    ax.tick_params(
        axis="y",
        labelsize=8
    )

    style_axis(ax)

plt.tight_layout()

hydro_path = os.path.join(
    output_dir,
    "boxplots_hidrologia.png"
)

plt.savefig(
    hydro_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================================================
# 3) BOXPLOTS DE PRECIPITACION - INUMET
# =========================================================

precip_cols = []

for st in inumet_stations:

    candidates = [
        c for c in df.columns
        if c.endswith(f" {st}")
        and "Precipitacion (P) [mm]" in c
    ]

    if len(candidates) == 1:
        precip_cols.append(candidates[0])

precip_data = [
    df[c].dropna().values
    for c in precip_cols
]

precip_labels = [
    c.replace(
        "Precipitacion (P) [mm] ",
        ""
    )
    for c in precip_cols
]

fig, ax = plt.subplots(
    figsize=(12, 5)
)

ax.boxplot(
    precip_data,
    tick_labels=precip_labels,
    widths=0.6,
    patch_artist=False,
    showfliers=True,
    flierprops=flier_style
)

ax.set_ylabel("Precipitación [mm]")

ax.tick_params(
    axis="x",
    labelrotation=45,
    labelsize=8
)

ax.tick_params(
    axis="y",
    labelsize=8
)

style_axis(ax)

plt.tight_layout()

precip_path = os.path.join(
    output_dir,
    "boxplots_precipitacion_inumet.png"
)

plt.savefig(
    precip_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================================================
# 4) BOXPLOTS DE VARIABLES METEOROLOGICAS - INIA
# =========================================================

inia_order = [
    "Evapotranspiracion Penman (ET) [mm] INIA Las Brujas",
    "Heliofania (Hel) [hs] INIA Las Brujas",
    "Humedad relativa media (HR) [%] INIA Las Brujas",
    "Radiacion solar (RS) [cal/cm2/dia] INIA Las Brujas",
    "Temperatura del aire media (TA) [ºC(24hs)] INIA Las Brujas",
    "Temperatura del aire maxima (TA) [ºC] INIA Las Brujas",
    "Temperatura del aire minima (TA) [ºC] INIA Las Brujas",
    "Velocidad del viento (VV) [2m/km/24h] INIA Las Brujas"
]

inia_cols = [
    c for c in inia_order
    if c in df.columns
]

n_vars = len(inia_cols)

ncols = 2
nrows = math.ceil(n_vars / ncols)

fig, axes = plt.subplots(
    nrows=nrows,
    ncols=ncols,
    figsize=(11, 3.6 * nrows)
)

axes = axes.flatten()

for i, col in enumerate(inia_cols):

    ax = axes[i]

    data = df[col].dropna().values

    short_name = col.replace(
        " INIA Las Brujas",
        ""
    )

    ax.boxplot(
        [data],
        tick_labels=["Las Brujas"],
        widths=0.5,
        patch_artist=False,
        showfliers=True,
        flierprops=flier_style
    )

    # Identificación de la variable de cada panel
    ax.set_title(
        wrap_title(short_name, width=28),
        fontsize=9
    )

    ax.tick_params(
        axis="x",
        labelsize=8
    )

    ax.tick_params(
        axis="y",
        labelsize=8
    )

    style_axis(ax)

# Apagar paneles sobrantes
for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.tight_layout()

inia_path = os.path.join(
    output_dir,
    "boxplots_meteorologia_inia.png"
)

plt.savefig(
    inia_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# =========================================================
# ARCHIVOS GENERADOS
# =========================================================

print("Archivos guardados:")
print(quality_path)
print(hydro_path)
print(precip_path)
print(inia_path)
```

```python
import pandas as pd
import numpy as np

path1 = './0-Datasets/dataset_feature_eng.parquet'
dataset_feature_eng = pd.read_parquet(path1)

# =========================================================
# VARIABLES OBJETIVO
# =========================================================
iet_lotico = "IET(PT) Lótico_Lamparelli [XSLH030]"
iet_lentico = "IET(PT) Léntico_Cunha [EPSE040]"

# =========================================================
# SPLIT TEMPORAL 80/20
# Solo se usará el 80% inicial para ranking + poda a efectos de asi evitar la fuga de datos!
# =========================================================
train_size = 0.8
split_idx = int(len(dataset_feature_eng) * train_size)

dataset_feature_eng_train80 = dataset_feature_eng.iloc[:split_idx].copy()

print(f"Filas totales: {len(dataset_feature_eng)}")
print(f"Filas usadas para selección de variables (80% inicial): {len(dataset_feature_eng_train80)}")
print(f"Filas reservadas fuera de esta selección: {len(dataset_feature_eng) - len(dataset_feature_eng_train80)}")
```

```python
# =========================================================
# RANKING DE CORRELACIONES PROMEDIO ABSOLUTO
# + FILTRADO POR UMBRAL
# Todo calculado SOLO en el 80% inicial
# =========================================================

def ranking_correlaciones_promedio_abs(df, target):
    # ---------------------------------------------
    # 1) Conversión numérica
    # ---------------------------------------------
    X = df.apply(pd.to_numeric, errors="coerce")
    y = pd.to_numeric(df[target], errors="coerce")

    # ---------------------------------------------
    # 2) Correlaciones absolutas
    # ---------------------------------------------
    pearson = X.corrwith(y, method="pearson").abs()
    spearman = X.corrwith(y, method="spearman").abs()
    kendall = X.corrwith(y, method="kendall").abs()

    promedio_abs = pd.concat([pearson, spearman, kendall], axis=1).mean(axis=1)

    tabla = pd.DataFrame({
        "Variable": promedio_abs.index,
        "Valor_abs_promedio": promedio_abs.values
    })

    # ---------------------------------------------
    # 3) Se eliminan explícitamente variables no deseadas
    # ---------------------------------------------
    # Se quita la propia variable target
    tabla = tabla[tabla["Variable"] != target]

    # quitar fósforos actuales (pero NO sus lags)
    tabla = tabla[
        ~(
            tabla["Variable"].str.contains("Fósforo Total \\(PT\\)", regex=True, na=False)
            & ~tabla["Variable"].str.contains("_lag", regex=False, na=False)
        )
    ]

    # quitar el otro IET y sus lags
    if target == iet_lotico:
        tabla = tabla[
            ~tabla["Variable"].isin([
                iet_lentico,
                f"{iet_lentico}_lag1",
                f"{iet_lentico}_lag2",
                f"{iet_lentico}_lag3"
            ])
        ]
    elif target == iet_lentico:
        tabla = tabla[
            ~tabla["Variable"].isin([
                iet_lotico,
                f"{iet_lotico}_lag1",
                f"{iet_lotico}_lag2",
                f"{iet_lotico}_lag3"
            ])
        ]

    # ---------------------------------------------
    # 4) Se ordena
    # ---------------------------------------------
    tabla = tabla.sort_values("Valor_abs_promedio", ascending=False).reset_index(drop=True)

    return tabla


# =========================================================
# SE CALCULAN RANKINGS SOLO EN EL 80% INICIAL
# =========================================================
corr_iet_lotico = ranking_correlaciones_promedio_abs(
    df=dataset_feature_eng_train80,
    target=iet_lotico
)

corr_iet_lentico = ranking_correlaciones_promedio_abs(
    df=dataset_feature_eng_train80,
    target=iet_lentico
)

print("\nCorrelaciones para IET lótico (calculadas solo en el 80% inicial):\n")
print(corr_iet_lotico.round(4))

print("\nCorrelaciones para IET léntico (calculadas solo en el 80% inicial):\n")
print(corr_iet_lentico.round(4))


# =========================================================
# FILTRADO POR RELEVANCIA
# =========================================================
umbral_corr = 0.5

corr_iet_lotico_filtrado = (
    corr_iet_lotico[corr_iet_lotico["Valor_abs_promedio"] > umbral_corr]
    .sort_values("Valor_abs_promedio", ascending=False)
    .reset_index(drop=True)
)

corr_iet_lentico_filtrado = (
    corr_iet_lentico[corr_iet_lentico["Valor_abs_promedio"] > umbral_corr]
    .sort_values("Valor_abs_promedio", ascending=False)
    .reset_index(drop=True)
)

print(f"\nVariables seleccionadas para IET lótico con umbral > {umbral_corr}:")
print(corr_iet_lotico_filtrado)

print(f"\nVariables seleccionadas para IET léntico con umbral > {umbral_corr}:")
print(corr_iet_lentico_filtrado)
```

```python
# =========================================================
# PODA DE COLINEALIDAD ENTRE PREDICTORES
# usando promedio(|Pearson|, |Spearman|, |Kendall|)
# Todo calculado SOLO en el 80% inicial
# =========================================================

def podar_colineales_promedio_abs(df_base, df_corr_filtrado, umbral_colinealidad=0.8):
    """
    df_base: dataset usado como base (aquí, solo el 80% inicial)
    df_corr_filtrado: DataFrame con columnas ['Variable', 'Valor_abs_promedio']
    umbral_colinealidad: umbral sobre promedio(|Pearson|, |Spearman|, |Kendall|)
    """

    variables = df_corr_filtrado["Variable"].tolist()

    # Subset numérico
    X = df_base[variables].apply(pd.to_numeric, errors="coerce")

    # Correlaciones absolutas entre predictores
    corr_pearson = X.corr(method="pearson").abs()
    corr_spearman = X.corr(method="spearman").abs()
    corr_kendall = X.corr(method="kendall").abs()

    # Promedio absoluto
    corr_promedio = (corr_pearson + corr_spearman + corr_kendall) / 3

    seleccionadas = []
    descartadas = []

    for var in df_corr_filtrado["Variable"]:
        es_colineal = False

        for sel in seleccionadas:
            corr_abs_prom = corr_promedio.loc[var, sel]

            if corr_abs_prom >= umbral_colinealidad:
                descartadas.append({
                    "Variable_descartada": var,
                    "Variable_conservada": sel,
                    "Corr_abs_promedio": corr_abs_prom
                })
                es_colineal = True
                break

        if not es_colineal:
            seleccionadas.append(var)

    df_seleccionadas = df_corr_filtrado[
        df_corr_filtrado["Variable"].isin(seleccionadas)
    ].reset_index(drop=True)

    df_descartadas = pd.DataFrame(descartadas)

    return df_seleccionadas, df_descartadas, corr_promedio


# =========================================================
# PODA POR COLINEALIDAD
# =========================================================
umbral_colinealidad = 0.8

# LÓTICO
corr_iet_lotico_final, corr_iet_lotico_descartadas, corr_pred_lotico = podar_colineales_promedio_abs(
    df_base=dataset_feature_eng_train80,
    df_corr_filtrado=corr_iet_lotico_filtrado,
    umbral_colinealidad=umbral_colinealidad
)

print(f"\nVariables finales para IET lótico tras poda por colinealidad promedio (|r̄| < {umbral_colinealidad}):")
print(corr_iet_lotico_final)

print(f"\nVariables descartadas por colinealidad para IET lótico:")
print(corr_iet_lotico_descartadas)


# LÉNTICO
corr_iet_lentico_final, corr_iet_lentico_descartadas, corr_pred_lentico = podar_colineales_promedio_abs(
    df_base=dataset_feature_eng_train80,
    df_corr_filtrado=corr_iet_lentico_filtrado,
    umbral_colinealidad=umbral_colinealidad
)

print(f"\nVariables finales para IET léntico tras poda por colinealidad promedio (|r̄| < {umbral_colinealidad}):")
print(corr_iet_lentico_final)

print(f"\nVariables descartadas por colinealidad para IET léntico:")
print(corr_iet_lentico_descartadas)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# FUNCIÓN: MATRIZ PROMEDIO ABSOLUTA
# =========================================================
def matriz_corr_promedio_abs(df_base, variables):
    X = df_base[variables].apply(pd.to_numeric, errors="coerce")

    corr_pearson = X.corr(method="pearson").abs()
    corr_spearman = X.corr(method="spearman").abs()
    corr_kendall = X.corr(method="kendall").abs()

    corr_promedio = (corr_pearson + corr_spearman + corr_kendall) / 3
    return corr_promedio


# =========================================================
# FUNCIÓN: HEATMAP TRIANGULAR
# =========================================================
def guardar_heatmap_triangular(
    corr_matrix,
    titulo,
    nombre_archivo,
    carpeta_salida="Figuras",
    figsize=(12, 10),
    cmap="viridis"
):
    os.makedirs(carpeta_salida, exist_ok=True)

    # Mostrar solo triángulo superior
    mask = np.tril(np.ones_like(corr_matrix, dtype=bool), k=-1)

    plt.figure(figsize=figsize)
    sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=True,
        fmt=".2f",
        cmap=cmap,
        vmin=0,
        vmax=1,
        square=True,
        linewidths=0.5,
        cbar_kws={"label": "Promedio abs(Pearson, Spearman, Kendall)"}
    )

    plt.title(titulo, fontsize=14)
    plt.xticks(rotation=45, ha="right", fontsize=11)
    plt.yticks(rotation=0, fontsize=11)
    plt.tight_layout()

    ruta_salida = os.path.join(carpeta_salida, nombre_archivo)
    plt.savefig(ruta_salida, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"Figura guardada en: {ruta_salida}")


# =========================================================
# LÓTICO: variables podadas + target
# =========================================================
variables_lotico_con_target = [iet_lotico] + corr_iet_lotico_final["Variable"].tolist()

corr_prom_lotico = matriz_corr_promedio_abs(
    df_base=dataset_feature_eng_train80,
    variables=variables_lotico_con_target
)

guardar_heatmap_triangular(
    corr_matrix=corr_prom_lotico,
    titulo="Correlación promedio absoluta - Lótico",
    nombre_archivo="heatmap_corr_promedio_lotico_con_target.png",
    carpeta_salida="Figuras",
    figsize=(12, 10),
    cmap="viridis"
)


# =========================================================
# LÉNTICO: variables podadas + target
# =========================================================
variables_lentico_con_target = [iet_lentico] + corr_iet_lentico_final["Variable"].tolist()

corr_prom_lentico = matriz_corr_promedio_abs(
    df_base=dataset_feature_eng_train80,
    variables=variables_lentico_con_target
)

guardar_heatmap_triangular(
    corr_matrix=corr_prom_lentico,
    titulo="Correlación promedio absoluta - Léntico",
    nombre_archivo="heatmap_corr_promedio_lentico_con_target.png",
    carpeta_salida="Figuras",
    figsize=(12, 10),
    cmap="viridis"
)
```
