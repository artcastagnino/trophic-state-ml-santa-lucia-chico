```python
import pandas as pd

path_dinacea_crudo = './0-Datasets/SL_CA_DINACEA_2011_2025.xlsx'

raw = pd.read_excel(path_dinacea_crudo, sheet_name=0, header=None)

# Fila 0: nombre de variable (viene con celdas combinadas => hay que "rellenar hacia la derecha")
vars_ = raw.iloc[0, 1:].ffill()

# Fila 1: estaciones/códigos
stations = raw.iloc[1, 1:]

# Datos: desde la fila 3 en adelante (0=vars, 1=station, 2=date)
data = raw.iloc[3:, 1:]

# Índice fecha: columna 0 desde la fila 3
dates = pd.to_datetime(raw.iloc[3:, 0], errors="coerce")

# Armar DataFrame limpio con columnas MultiIndex
data_dinacea_crudo = data.copy()
data_dinacea_crudo.index = dates
data_dinacea_crudo.index.name = "date"
data_dinacea_crudo.columns = pd.MultiIndex.from_arrays([vars_, stations], names=["variable", "station"])

# Convertir a numérico (por si hay strings sueltos)
data_dinacea_crudo = data_dinacea_crudo.apply(pd.to_numeric, errors="coerce")

data_dinacea_crudo.head()
```

```python
# Se seleccionan del dataset de calidad de agua las estaciones ubicadas en la cuenca de Paso Severino

cols = ["EPSE020", "EPSE040", "EPSE030", "XSLH030", "XSLH020", "XSLH010"]
data_dinacea_crudo = data_dinacea_crudo.loc[:, data_dinacea_crudo.columns.get_level_values(1).isin(cols)].copy()

data_dinacea_crudo.head()
```

```python
import pandas as pd

# lectura del dataset crudo de DINAGUA
path_dinagua_crudo = './0-Datasets/SL_HIDRO_DINAGUA_1980_2025.xlsx'
raw = pd.read_excel(path_dinagua_crudo, sheet_name=0, header=None)

# detectar filas "station" y "date" (en la primera columna)
col0 = raw.iloc[:, 0].astype(str).str.strip().str.lower()
station_row = col0[col0.eq("station")].index[0]
date_row    = col0[col0.eq("date")].index[0]

# asumimos que la fila anterior a "station" contiene el "grupo/variable" (celdas combinadas -> ffill)
var_row = station_row - 1

vars_ = raw.iloc[var_row, 1:].astype(str).str.strip()
vars_ = vars_.replace({"nan": None, "None": None}).ffill()

stations = raw.iloc[station_row, 1:].astype(str).str.strip()

# los datos empiezan después de la fila "date"
data_start = date_row + 1

dates = pd.to_datetime(raw.iloc[data_start:, 0], errors="coerce")
data  = raw.iloc[data_start:, 1:]

data_dinagua_crudo = data.copy()
data_dinagua_crudo.index = dates
data_dinagua_crudo.index.name = "date"

data_dinagua_crudo.columns = pd.MultiIndex.from_arrays(
    [vars_, stations],
    names=["variable", "station"]
)

# convertir a numérico
data_dinagua_crudo = data_dinagua_crudo.apply(pd.to_numeric, errors="coerce")

data_dinagua_crudo.head()
```

```python
# Se seleccionan del dataset de dinagua el caudal y nivel de agua de la estación Florida (Puente Ruta 5)

idx = pd.IndexSlice
est = "Florida (Puente Ruta 5)"

data_dinagua_crudo = data_dinagua_crudo.loc[:, idx[["Caudal (Q) [m3/s]", "Nivel de agua (h) [m]"], [est]]]
data_dinagua_crudo.head()
```

```python
import pandas as pd

# lectura del dataset crudo de INIA
path_inia_crudo = './0-Datasets/SL_MET_INIA_1980_2025.xlsx'
raw = pd.read_excel(path_inia_crudo, sheet_name=0, header=None)

# detectar filas "station" y "date" (en la primera columna)
col0 = raw.iloc[:, 0].astype(str).str.strip().str.lower()
station_row = col0[col0.eq("station")].index[0]
date_row    = col0[col0.eq("date")].index[0]

# la fila anterior a "station" suele contener el grupo/variable (celdas combinadas -> ffill)
var_row = station_row - 1

vars_ = raw.iloc[var_row, 1:].astype(str).str.strip()
vars_ = vars_.replace({"nan": None, "None": None}).ffill()

stations = raw.iloc[station_row, 1:].astype(str).str.strip()

# los datos empiezan después de la fila "date"
data_start = date_row + 1
dates = pd.to_datetime(raw.iloc[data_start:, 0], errors="coerce")
data  = raw.iloc[data_start:, 1:]

data_inia_crudo = data.copy()
data_inia_crudo.index = dates
data_inia_crudo.index.name = "date"

data_inia_crudo.columns = pd.MultiIndex.from_arrays(
    [vars_, stations],
    names=["variable", "station"]
)

# convertir a numérico
data_inia_crudo = data_inia_crudo.apply(pd.to_numeric, errors="coerce")

data_inia_crudo.head()
```

```python
# Se realiza el pasaje a valores promedios diarios mensuales (un valor por mes)

dinagua_mensual_crudo = data_dinagua_crudo.resample('ME').mean()

# Se ajusta el índice para asignar la fecha al último día de cada mes

dinagua_mensual_crudo.index = dinagua_mensual_crudo.index.to_period('M').to_timestamp('M')

dinagua_mensual_crudo.head()
```

```python
# Se realiza el pasaje a valores promedios diarios mensuales (un valor por mes)

inumet_mensual_crudo = data_inumet_crudo.resample('ME').mean()

# Se asegura que el índice quede exactamente como "último día"
inumet_mensual_crudo.index = inumet_mensual_crudo.index.to_period('M').to_timestamp('M')

inumet_mensual_crudo.head()
```

```python
# Se realiza el pasaje a valores promedios diarios mensuales (un valor por mes)

inia_mensual_crudo = data_inia_crudo.resample('ME').mean()

# Se ajusta el índice para asignar la fecha al último día de cada mes

inia_mensual_crudo.index = inia_mensual_crudo.index.to_period('M').to_timestamp('M')

inia_mensual_crudo.head()
```

```python
# Se asegura que todos los sets tengan el mismo índice de fecha

data_dinacea_crudo.index = pd.to_datetime(data_dinacea_crudo.index)
dinagua_mensual_crudo.index = pd.to_datetime(dinagua_mensual_crudo.index)
inumet_mensual_crudo.index = pd.to_datetime(inumet_mensual_crudo.index)
inia_mensual_crudo.index = pd.to_datetime(inia_mensual_crudo.index)

# Se combinan los datasets

dataset_completo_crudo = pd.concat([data_dinacea_crudo, dinagua_mensual_crudo, inumet_mensual_crudo, inia_mensual_crudo], axis=1)

dataset_completo_crudo.head()
```

```python
dataset_completo_crudo = dataset_completo_crudo.loc["2014-01-01":"2022-12-31"].copy()

dataset_completo_crudo.head()
```

```python
dataset_completo_crudo.columns = [' '.join(col).strip() for col in dataset_completo_crudo.columns.values]
dataset_completo_crudo.head()
```

```python
# Se obtiene un dataset mensual de extremos de caudal a partir del dataset crudo

dinagua_mensual_extremos_crudo = (
    data_dinagua_crudo[[('Caudal (Q) [m3/s]', 'Florida (Puente Ruta 5)')]]
    .resample('ME')
    .agg(['max', 'min'])
)

# Se ajusta el índice al último día de cada mes
dinagua_mensual_extremos_crudo.index = dinagua_mensual_extremos_crudo.index.to_period('M').to_timestamp('M')

dinagua_mensual_extremos_crudo.columns = [
    'Caudal Maximo (Qmax) [m3/s] Florida (Puente Ruta 5)',
    'Caudal Minimo (Qmin) [m3/s] Florida (Puente Ruta 5)'
]

# Se selecciona el intervalo de trabajo
dinagua_mensual_extremos_crudo = dinagua_mensual_extremos_crudo.loc["2014-01-01":"2022-12-31"]

dinagua_mensual_extremos_crudo.head()
```

```python
# Se obtiene un dataset mensual de precipitaciones máximas a partir del dataset crudo

inumet_mensual_extremos_crudo = (
    data_inumet_crudo
    .resample('ME')
    .max()
)

# Se ajusta el índice al último día de cada mes
inumet_mensual_extremos_crudo.index = inumet_mensual_extremos_crudo.index.to_period('M').to_timestamp('M')

inumet_mensual_extremos_crudo.columns = [
    f'Precipitacion maxima (Pmax) [mm] {station}'
    for variable, station in inumet_mensual_extremos_crudo.columns
]

# Se selecciona el intervalo de trabajo
inumet_mensual_extremos_crudo = inumet_mensual_extremos_crudo.loc["2014-01-01":"2022-12-31"]

inumet_mensual_extremos_crudo.head()
```

```python
# Se asegura que todos los sets tengan el mismo índice de fecha

dataset_completo_crudo.index = pd.to_datetime(dataset_completo_crudo.index)
dinagua_mensual_extremos_crudo.index = pd.to_datetime(dinagua_mensual_extremos_crudo.index)
inumet_mensual_extremos_crudo.index = pd.to_datetime(inumet_mensual_extremos_crudo.index)

# Se combinan los datasets
dataset_completo_crudo = pd.concat(
    [
        dataset_completo_crudo,
        dinagua_mensual_extremos_crudo,
        inumet_mensual_extremos_crudo
    ],
    axis=1
)

dataset_completo_crudo.head()
```
