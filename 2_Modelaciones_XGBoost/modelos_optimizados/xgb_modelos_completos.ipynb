### Lógica general de los datasets de correlación

| Nombre genérico | Explicación breve |
|---|---|
| `correlacion_top3` | Las 3 variables con mayor correlación absoluta. |
| `correlacion_top4` | Las 4 variables con mayor correlación absoluta. |
| `correlacion_meteorologia` | Solo las variables meteorológicas que aparecen dentro del ranking de correlación. |
| `correlacion_todos_sinP` | Todas las variables correlacionadas consideradas, excepto las de fósforo. |
| `correlacion_todos` | Todas las variables correlacionadas. |

### Lógica general de los datasets de relación física

| Nombre genérico | Explicación breve |
|---|---|
| `relfis_agua` | Todas las variables de calidad de agua seleccionadas por criterio físico, excluyendo caudales. |
